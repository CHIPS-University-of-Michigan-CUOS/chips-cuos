```python
import numpy as np
import scipy.constants as const
```


* critical density
$$n_{c}[cm^{-3}] = \frac{m_{e}\epsilon_{0}}{e^{2}}\omega_{0}^2 = 
\frac{4\pi^{2}m_{e}\epsilon_{0}c^2}{e^{2}}\frac{1}{\lambda_{0}^{2}}
= 1.114854\times10^{21}\frac{1}{\lambda_{0}[\mu m]^2}$$
* For laser wavelength 800nm, $$n_{c} = 1.7419597\times10^{21}cm^{-3} $$

```python
me = const.electron_mass
epsilon = const.epsilon_0
c0 = const.c
q0 = const.elementary_charge
```

```python
def criticalDensity(wavelen):
    '''wavelen in um
       density in cm^-3'''
    return 4*np.pi**2*me*epsilon*c0**2*1e6/q0**2/wavelen**2
nc = criticalDensity(0.8)
nc
```

```python
nc = criticalDensity(3.9)
```

## Normalized to laser units
* time normalized to $$t_{0}= 2\pi/\omega_{0}= T_{0}$$
* space normalized to $$x_{0} = 2\pi c/\omega_{0}=\lambda_{0}$$

Therefore, $$\omega_{p} = \omega_{0} / 2\pi$$

$$\omega_{0} = 2\pi\omega_{p}$$

$$n_{0} = n_{c}/4\pi^2 = 4.4124355\times10^{19}[cm^{-3}]$$

```python
wavelen = 3.9 # in um
omega0 =2*np.pi*c0/(wavelen*1e-6) # in rad/s
t0 = wavelen*1e-6/c0/1e-15 # in fs
x0 = wavelen # in um
n0 = nc/4/np.pi**2
```

## Courant condition:
$$c^{2}dt^{2}<\frac{1}{1/dx^{2}+1/dy^{2}}$$
$$dt^{2}<\frac{1}{c^{2}/dx^{2}+ c^{2}/dy^{2}}$$
define:
$$dt = 0.995 \sqrt{\frac{1}{c^{2}/dx^{2}+ c^{2}/dy^{2}}}$$

```python
def dt(dx,dy):
    '''dt in s
       dx, dy in um'''
    return 0.995*np.sqrt(1/(c0**2/(dx*1e-6)**2 + c0**2/(dy*1e-6)**2))
dt(x0/32, x0/16)/1e-15/t0
```
